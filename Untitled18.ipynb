```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import matplotlib.pyplot as plt
```

```python
# Dataset
data = {
    'Fever':            ['NO','YES','YES','YES','YES','NO','YES','YES','NO','YES','NO','NO','NO','YES'],
    'Cough':             ['NO','YES','YES','NO','YES','YES','NO','NO','YES','YES','YES','YES','YES','YES'],
    'Breathing_issues':  ['NO','YES','NO','YES','YES','NO','YES','YES','YES','NO','NO','YES','YES','NO'],
    'Infected':          ['NO','YES','NO','YES','YES','NO','YES','YES','YES','YES','NO','YES','NO','NO']
}
df1 = pd.DataFrame(data)

le = LabelEncoder()
df1_enc = df1.apply(le.fit_transform)

X1 = df1_enc.drop('Infected', axis=1)
y1 = df1_enc['Infected']

clf1 = DecisionTreeClassifier(criterion='entropy', random_state=0)
clf1.fit(X1, y1)

print("Predictions:", clf1.predict(X1).tolist())
print("Actual:     ", y1.tolist())
print("Accuracy:", accuracy_score(y1, clf1.predict(X1)))

plt.figure(figsize=(10, 6))
plot_tree(clf1, feature_names=X1.columns, class_names=['NO', 'YES'], filled=True)
plt.title("Decision Tree - Infection Dataset")
plt.show()
```

```python
from sklearn.datasets import load_breast_cancer

cancer = load_breast_cancer()
X2 = pd.DataFrame(cancer.data, columns=cancer.feature_names)
y2 = cancer.target  # 0 = malignant, 1 = benign

X_train, X_test, y_train, y_test = train_test_split(X2, y2, test_size=0.2, random_state=42)

clf2 = DecisionTreeClassifier(criterion='entropy', max_depth=4, random_state=42)
clf2.fit(X_train, y_train)

y_pred = clf2.predict(X_test)

print("Train Accuracy:", clf2.score(X_train, y_train))
print("Test Accuracy:", clf2.score(X_test, y_test))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred, target_names=['malignant', 'benign']))

plt.figure(figsize=(20, 10))
plot_tree(clf2, feature_names=cancer.feature_names, class_names=['malignant', 'benign'], filled=True, max_depth=2)
plt.title("Decision Tree - Breast Cancer Dataset")
plt.show()
```
